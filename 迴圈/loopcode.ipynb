# Loop

Loop指令可以讓程式碼依我的條件不斷地執行，直到條件滿足為止。

```python
# 底下的程式碼會印出 0-9
for i in range(10):
    print(i)
```

```python
list(range(10))
```

```python
# 底下的程式碼會印出 0-9
for i in range(10):
    print(i+1)
```

# Example

計算1+2+3...+10

sum=0
i=0| i+1=1| sum=sum+(i+1)-->sum=(0)+1

i=1| i+1=2| sum=sum+(i+1)-->sum=(0+1)+2

i=2| i+1=3| sum=sum+(i+1)-->sum=(1+2)+3

i=3| i+1=4| sum=sum+(i+1)-->sum=(1+2+3)+4

```python
sum=0
for i in range(10):
    sum=sum+(i+1)

print(sum)
```

# Range

python:

-懶惰是一種美德

-由0開始

-包含頭，不包含尾

```python
print(list(range(5)))
#區間，由0開始，到5-1結束
print(list(range(0,5)))
#區間，由1開始，到5-1結束
print(list(range(1,5)))
#區間，由1開始，到5-1結束，每次跳2個
print(list(range(1,5,2)))
```
